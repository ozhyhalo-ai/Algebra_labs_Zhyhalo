### 1. Властивості циклічних груп

Для наступних порядків $n \in \{100, 173, 530, 746, 911, 1024\}$ розгляньте циклічну групу $G = \mathbb{Z}_n$ (або $C_n$).

1. Знайдіть усі дільники $d$ порядку $n$.<br>
2. Для кожного дільника $d$ обчисліть кількість елементів $a \in G$ таких, що $\text{ord}(a) = d$.<br>
3. Порівняйте отриману кількість з теоретичним значенням $\phi(d)$, де $\phi$ --- функція Ойлера.<br>
4. Знайдіть усі підгрупи групи $G$.<br>
5. Обчисліть загальну кількість усіх підгруп.<br>
6. Порівняйте отриману кількість з теоретичним значенням $\tau(n)$, де $\tau$ -- функція кількості дільників числа $n$.



```sage
#G = CyclicPermutationGroup(24)
#subgroups = G.subgroups()

N = [100, 173, 530, 746, 911, 1024]
for n in N:
    print(f"\nЦиклічна група порядку n = {n}")
    G = CyclicPermutationGroup(n)
    
    # 1.
    d_list = divisors(n) # усі дільники
    print(f"\n1. Дільники числа {n}: {d_list}")
    
    # 2.
    print("\n2.та3. Порівняння з функцією Ойлера (фі):")
    for d in d_list:
        count = sum(1 for a in G if a.order() == d)
        
        # 3.
        # Теоретичне значення за функцією Ойлера
        euler_val = euler_phi(d)
        print(f"    Для d={d:4} Знайдено елементів: {count:4} За Ойлером: {euler_val:4}")
        
    # 4.
    subs = G.subgroups()
    print(f"\n4. Усього {len(subs)} груп")
    
    # 5.
    total_subs = len(subs)
    tau_val = number_of_divisors(n) # функція кількості дільників числа n
    print("\n5. Порівняння загальної кількості підгруп:")
    print(f"    Фактична кількість підгруп: {total_subs}")
    print(f"    Теоретична кількість (тау): {tau_val}")
    print(f"\n")
```

### 2. Порядки елементів в діедральних групах

Для наступних значень $n\in\{5,6,10,12,16,17,20\}$ розгляньте діедральну групу $D_n$.   

1. Знайдіть кількість елементів порядку $d=2$ в групі $D_n$.
2. Для кожного дільника $d>2$ порядку $|D_n|=2n$ знайдіть кількість елементів порядку $d$ у групі $D_n$.<br>

Проаналізуйте результати та дайте відповіді на запитання:

1. Скільки елементів порядку $d=2$ в діедральній групі $D_n$?
2. Скільки елементів порядку $d>2$ в діедральній групі $D_n$?
3. Чи для кожного дільника $d<2n$ порядку групи $D_n$ існує елемент порядку $d$?



```sage
N = [5, 6, 10, 12, 16, 17, 20]
for n in N:
    print(f"\nДіедральна група D_{n} (загальний порядок {2*n})")
    D = DihedralGroup(n)
    order_D = 2*n
    
    d_list = divisors(order_D)
    
    # 1. d=2
    count_2 = sum(1 for a in D if a.order() == 2)
    print(f"1. Кількість елементів порядку d=2: {count_2}")
    
    # 2. d>2
    print("2. Кількість елементів для дільників d>2:")
    for d in d_list:
        if d>2:
            count_d = sum(1 for a in D if a.order() == d)
            print(f"    Для d={d:2} Знайдено елементів: {count_d:2}")
```

1. _Скільки елементів порядку_ $d=2$ _в діедральній групі_ $D_n$_?_ 

Кількість залежить від парності $n$:

Якщо $n$ непарне, то таких елементів рівно $n$. Це виключно відбиття.

Якщо $n$ парне, то таких елементів $n+1$. До $n$ відбиттів додаються ще поророт на $π$ ($180º$), який за два кроки повертає фігуру в початковий стан.",

2. _Скільки елементів порядку_ $d>2$ _в діедральній групі_ $D_n$_?_

Такі елементи можуть бути виключно серед обертань, тому їхня кількість підпорядковується правилам циклічних груп:

Якщо $d$ є дільником самого числа $n$, то кількість таких елементів дорівнює функції Ойлера.

Якщо $d$ не є дільником числа $n$, то таких елементів $0$.

3. _Чи для кожного дільника_ $d<2n$ _порядку групи_ $D_n$ _існує елемент порядку_ $d$ _?_

Ні, не для кожного.

Контрприклад: у групі $D_{10}$ є дільник $4$. Але елементів порядку $4$ знаєдено програмою $0$.

### 3. Дискретний логарифм

Розглянемо групу $\mathbb{Z}_n^{*}$ для $n=1102158078129785185997827239806$.

1. Знайдіть факторизацію модуля $n$.
2. Перевірте, що група є циклічною, та знайдіть її порядок.
3. Знайдіть найменше натуральне число $a \in \mathbb{N}$ таке, що елемент $\overline{a} \in \mathbb{Z}_n^{*}$ є твірним елементном групи.
4. Перевірте, що порядок елемента $\overline{a}$ дорівнює $\phi(n)$.
5. Виберіть три випадкові елементи $g_1, g_2, g_3 \in \mathbb{Z}_n^{*}$.
6. Для кожного елемента $g \in \{g_1, g_2, g_3\}$ знайдіть дискретний логарифм за модулем $n$, тобто таке число $k$, що $g=\overline{a}^k$ в $\mathbb{Z}_n^{*}$.



```sage
n=1102158078129785185997827239806
Z = Integers(n)
print(f"Мультиплікативна група Zn^* для n={n}")

# 1. Факторизація модуля
fact = factor(n)
print(f"\n1. Факторизація n: {fact}")

# 2. Перевірка порядку групи
order_Z = euler_phi(n)
print(f"\n2. Порядок групи: {order_Z}")

# 3. Найменше число, яке є твірним ел.
a_val = primitive_root(n)
print(f"3. Найменший твірний елемент: a = {a_val}")

# 4. Перевірка порядку ел. a
a_mod = Z(a_val)
order_a = a_mod.multiplicative_order()
print(f"\n4. Порядок елемента a: {order_a}")
if order_a == order_Z:
    print("Він збігається з функцією Ойлера")
else:
    print("Він НЕ збігається з функцією Ойлера")
    
# 5. 3 випадкові елемента
g_list = []
while len(g_list) < 3:
    rang_g = Z.random_element()
    if gcd(rang_g, n) == 1:
        g_list.append(rang_g)
print(f"\n5. Три випадкові елементи:")
for g in (g_list):
    print(g)
    
# 6. Дискретний логарифм
print(f"\n6. Дискретні логарифми:")
for g in (g_list):
    k = g.log(a_mod)
    print("k =", k)
    
    # Перевірка
    check = (a_mod^k == g)
    print("Перевірка a^k == g: ", check)
```

### 4. Одна лінійна група

Розглянемо підгрупу $G$ групи $GL_3(\mathbb{Z})$, породжену двома матрицями:
$$G = \left\langle \begin{pmatrix} 1 & 0 & 1 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix}, \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 1 \\ 0 & 0 & 1 \end{pmatrix} \right\rangle.$$
Дайте відповідь на наступні питання, обґрунтовуючи відповіді за допомогою обчислень у SageMath:

1. Чи є група $G$ скінченною?
2. Чи є група $G$ абелевою?
3. Які порядки мають елементи групи $G$?
4. Який вигляд мають елементи групи $G$?
5. Чи є група $G$ циклічною?
6. Якій відомій групі ізоморфна група $G$?



```sage
from random import randint # щоб генерувати рандомно числа для демонстрації загального вигляду будь-якого елемента групи

# створюємо базові матриці над кільцем цілих чисел (ZZ)
A = matrix(ZZ, [[1, 0, 1], [0, 1, 0], [0, 0, 1]])
B = matrix(ZZ, [[1, 0, 0], [0, 1, 1], [0, 0, 1]])

print("Генератор групи G:")
print("Матриця A: \n", A)
print("\nМатриця B:\n", B)

# Скінченність та порядки ел.
print("\nДослідження на скінченність та порядки елементів:")
print("A^1 =\n", A^1)
print("\nA^5 =\n", A^5)
print("\nB^10 =\n", B^10)

# Перевіряємо абелевість
print("\nПеревірка на абелевість:")
print("A * B =\n", A*B)
print("\nB * A=\n", B*A)
print(f"\nЧи збігаються вони? {A * B == B * A}\n")

# Загальний вигляд будь-якого ел.
x = randint(-20, 20)
y = randint(-20, 20)
print(f"\nЗагальний вигляд елемента для випадкових x={x} та y={y}:")
print(f"A^({x}) * B^({y}) =\n", A^x * B^y)
```

1. _Чи є група_ $G$ _скінченною?_

Ні, група не є скінченною. Обчислення показують, що при піднесенні генераторів до степення значення у третьому стовпці лінійно зростають (або спадають) до нескінченності, не утворюючи замкненого циклу.

2. _Чи є група_ $G$ _абелевою?_

Так, група $G$ є абелевою. Програмана перевірка підтверджує, що генератори комутують: $A•B=B•A$. Оскільки базові елементи незалежні від порядку множення, то і вся група є комутативною.

3. _Які порядки мають елементи групи_ $G$_?_ 

Усі елементи (за винятком одиничної матриці, яка має порядок $1$) мають нескінченний порядок. Матриці ніколи не повертаються до початкового стану при піднесенні до жодного ненульового степення.

4. _Який вигляд мають елементи групи_ $G$_?_

Будь-який елемент є комбінацією $A^x•B^y$ і має вигляд верхньотрикутної матриці з одиницями по головній ліагоналі та цідими числами $x$ і $y$ у третьому стовпці:

\begin{pmatrix}1 & 0 & x \\ 0 & 1 & y \\ 0 & 0 & 1 \end{pmatrix} 

5. _Чи є група_ $G$ _циклічною?_

Ні, група не є циклічною. Для охоплення всіх можливих комбінацій значень $x$ та $y$ необхідні мінімум два незалежні генератори, тоді як циклічна грпа має породжуватися лише одним елементом.

6. _Якій відомій групі ізоморфна група_ $G$_?_

Група ізоморфна прямій сумі двох груп цілих чисел $\mathbb{Z}×\Z$ (або $\mathbb{Z}^2$). Множення матриць у цій підгрупі повністю еквівалентне звичайному покомпонентному додаванню векторів $(x_1,y_1)+(x_2,y_2)=(x_1+x_2,y_1+y_2)$.

### 5*. Циклічність груп $\mathbb{Z}_n^{*}$

Для кожного $n=2, 3, 4, \ldots, 256$ виконайте наступні кроки, представляючи результати у вигляді таблиці:

1. Знайдіть факторизацію числа $n$.
2. Перевірте, чи є група $\mathbb{Z}_n^{*}$ циклічною.
3. Для циклічних груп $\mathbb{Z}_n^{*}$ знайдіть найменше $a \in \mathbb{N}$ таке, що $\overline{a}$ є твірним групи.
4. Сформулюйте гіпотезу: для яких $n \in \mathbb{N}$ група $\mathbb{Z}_n^{*}$ є циклічною.



```sage
# Друкуємо шапку таблиці
print(f"{'n':<4} | {'Факторизація':<15} | {'Циклічна?':<10} | {'Твірний (a)':<5}")
print("-" * 50)

for n in range(2, 257):
    # 1. Факторизація
    fact = factor(n)
    
    # 2. та 3. Циклічність та твірний елемент
    try:
        a = primitive_root(n)
        is_cyclic = 'True'
        a_str = str(a)
    except ValueError:
        is_cyclic = 'False'
        a_str = '-'
        
    print(f"{n:<4} | {str(fact):<15} | {is_cyclic:<10} | {a_str:<5}")
```

<span style='font-size:medium'>_Гіпотеза:_</span>

Група $\mathbb{Z_n^*}$ є циклічною тоді і тільки тоді, коли модуль $n$ має один із таких чотирьох виглядів:

- $n=2$
- $n=4$
- $n=p^k$ (де $p$ - будь-яке просте число, а $k≥1$)
- $n=2•p^k$ (де $p$ - непарне просте число)

У всіх інших випадках група не буде циклічною.

### 6*. Числа Фібоначчі

Розглянемо підгрупу $H$ групи $GL_2(\mathbb{Z})$, породжену двома матрицями:
$$H = \left\langle \begin{pmatrix} 1 & -1 \\ -1 & 2 \end{pmatrix}, \begin{pmatrix} 3 & 2 \\ 2 & 1 \end{pmatrix} \right\rangle.$$
Дайте відповідь на наступні питання, обґрунтовуючи відповіді за допомогою обчислень у SageMath: 

1. Чи є група $H$ скінченною?
2. Чи є група $H$ абелевою?
3. Які порядки мають елементи групи $H$?
4. Який вигляд мають елементи групи $H$?
5. Чи є група $H$ циклічною? Якщо так, то знайдіть її твірний елемент.



```sage
A = matrix(ZZ, [[1, -1], [-1, 2]])
B = matrix(ZZ, [[3, 2], [2, 1]])

print("Генератори групи H:")
print("Матриця A:\n", A)
print("\nМатриця B:\n", B)

# 2. Перевіряємо абелевість
print("\nПеревірка на абелевість (комутативність):")
print("A * B =\n", A * B)
print("\nB * A =\n", B * A)
print(f"\nЧи збігаються вони? {A * B == B * A}")

# Знайдемо твірний елемент (матрицю Фібоначчі F)
F = A * B
print("\nМатриця F (добуток A і B):\n", F)

# 4. Досліджуємо вигляд елементів 
print("\nДосліджуємо вигляд елементів (числа Фібоначчі):")
for i in range(1, 7):
    print(f"F^{i} =\n{F^i}\n")
    
# 5. Перевірка, чи F генерує початкові матриці
print("\nПеревірка генерації:")
print(f"Чи дорівнює F^(-2) матриці A? {F^(-2) == A}")
print(f"Чи дорівнює F^3 матриці B? {F^3 == B}")
```

1. _Чи є група_ $H$ _скінченною?_

Ні. При піднесенні до степення числа в матриці генерують послідовність Фібоначчі, яка зростає до нескінченності.

2. _Чи є група_ $H$ _абелевою?_

Так. Добуток $A•B$ дорівнює $B•A$. Це логічно, оскільки обидві матриці є просто степеннями однієї і тієїж матриці $F$, а $F^x•F^y=F^{x+y}=F^y•F^x$.

3. _Які порядки мають елементи групи_ $H$_?_

Усі елементи (крім одиничної матриці) мають нескінченний порядок, оскільки числа Фібоначчі зростають і ніколи не повертаються до нульв чи одиниць (не утворюють циклу).

4. _Який вигляд мають елементи групи_ $H$_?_

Будь-який елемент групи має вигляд

\begin{pmatrix} F_{k+1} & F_k \\ F_k & F_{k-1} \end{pmatrix} , 
де $F_k$ - це $k$-те число Фібоначчі.

5. Чи є група $H$ циклічною?

Так. Вся група породжується лише одним твірним елементом. Цей генератор - матриця $F = \begin{pmatrix} 1 & 1 \\ 1 & 0 \end{pmatrix}$, яку було отримано множенням $A•B$.
