### 1. Класи суміжності та нормальні підгрупи

Розглянемо групу $S_4$ та її підгрупи:<br>
$\quad$ (a) $H_1=\langle (1,2,3,4)\rangle$;<br>
$\quad$ (b) $H_2=\langle (1,3), (1,2,3,4)\rangle$;<br>
$\quad$ (c) $H_3=K_4$.

Для кожної підгрупи $H$ виконайте наступні кроки:
1. Обчисліть порядок підгрупи $H$ та її індекс у групі $[S_4 : H]$.
2. Знайдіть ліві та праві класи суміжності підгрупи $H$ у $S_4$.
3. Визначте, чи є $H$ нормальною підгрупою $S_4$.
4. Якщо $H$ є нормальною, то побудуйте факторгрупу $G/H$, її таблицю Келі та визначіть структуру.
5. Якщо $H$ не є нормальною, то знайдіть такий елемент $a\in S_4$, що $aH\neq Ha$.


```sage
G = SymmetricGroup(4)

H1 = G.subgroup([G("(1,2,3,4)")])
H2 = G.subgroup([G("(1,3)"), G("(1,2,3,4)")])
H3 = G.subgroup([G("(1,2)(3,4)"), G("(1,3)(2,4)")]) # K4 - група Кляйна

subgroups = [
    ("H1 = <(1, 2, 3, 4)>", H1),
    ("H2 = <(1, 3), (1, 2, 3, 4)>", H2),
    ("H3 = K4", H3)
]

for name, H in subgroups:
    print(f"Аналіз підгрупи {name}\n")
    
    # 1. Обчислення порядку та індексу
    order_G = G.order()
    order_H = H.order()
    index_H = order_G // order_H
    print(f"\n1. Порядок H: {order_H}")
    print(f"   Індекс [S4 : H] = {order_G} / {order_H} = {index_H}")
    
    # 2. Знаходження лівих та правих класів суміжності (ВРУЧНУ)
    left_cosets = []
    right_cosets = []
    
    all_cosets_info = [] # тут збурігаємо елемент і його класи
    
    for g in G:
        l_coset = []
        r_coset = []
        for h in H:
            l_coset.append(g * h) # множимо зліва
            r_coset.append(h * g) # множимо справа
            
        # Сортуємо класи за їх текстовим виглядом (щоб мали елементи в однакововму порядку)
        l_coset.sort(key=str)
        r_coset.sort(key=str)
        
        all_cosets_info.append((g, l_coset, r_coset))
        
        if l_coset not in left_cosets:
            left_cosets.append(l_coset)
        if r_coset not in right_cosets:
            right_cosets.append(r_coset)
           
    print(f"\n2. Класи суміжності:")
    print(f"   Знайдено лівих класів: {len(left_cosets)} (має збігатися з індексом {index_H})")
    print(f"   Знайдено правих класів: {len(right_cosets)} (має збігатися з індексом {index_H})")
    
    # 3. Перевірка на нормальність
    is_normal_func = H.is_normal(G)
    # (ВРУЧНУ)
    is_normal_manual = True
    traitor = None
    traitor_diff_element = None
    traitor_l_coset = None 
    traitor_r_coset = None
    
    for g, l_coset, r_coset in all_cosets_info:
        if l_coset != r_coset:
            is_normal_manual = False
            for i in l_coset:
                if i not in r_coset:
                    traitor = g
                    traitor_diff_element = i
                    traitor_l_coset = l_coset # Зберігаємо класи саме для цього g
                    traitor_r_coset = r_coset
                    break 
            break 
            
    print(f"\n3. Чи є H нормальною підгрупою S4?")
    print(f"   За вбудованою функцією: {is_normal_func}")
    print(f"   За ручною перевіркою (aH == Ha): {is_normal_manual}")
    
    # 4 та 5. Дії залежно від нормальності
    if is_normal_manual:
        print(f"\n4. Оскільки H - нормальна, будуємо факторгрупу G/H.")
        Q = G.quotient(H)
        print("\n   Таблиця Келі факторгрупи:")
        print(Q.cayley_table(names='letters'))
        print(f"\n   Структура факторгрупи G/H: {Q.structure_description()}")
        
    else:
        print(f"\n5. Оскільки H НЕ є нормальною, шукаємо елемент a є S4, для якого aH != Ha.")
        print(f"   Знайдено елемент a = {traitor}")
        print(f"   Елемент {traitor_diff_element} є у лівому класі, але ВІДСУТНІЙ у правому.")
        print(f"   Лівий клас aH: {traitor_l_coset}")
        print(f"   Правий клас Ha: {traitor_r_coset}")
        
    print("\n")
    print("---------------------------------------")
    print("\n")
```

### 2. Нормальні підгрупи в діедральних групах

Розгляньте діедральні групи $D_n$ для значень $n\in\{3,4,\dots,30\}$ і виконайте такі кроки:

2. Визначте кількість нормальних підгруп у $D_n$.
3. Складіть перелік порядків усіх нормальних підгруп.
4. Визначте алгебраїчну структуру кожної нормальної підгрупи.
5. Знайдіть закономірність і виведіть гіпотезу про кількість та структуру нормальних підгруп в $D_n$.
6. Сформулювати гіпотезу: Скільки нормальних підгруп в $D_n$ та яка їхня структура? Спрогнозувати кількість нормальних підгруп у $D_{470448}$ без виконання обчислень. 

```sage
for n in range(3, 31):
    G = DihedralGroup(n)
    
    # 1. Знаходимо всі нормальні підгрупи за допомогою вбудованої функції
    #normal_subs = G.normal_subgroups()
    actual_count = len(normal_subs)
    
    # РУЧНИЙ ПОШУК
    all_subs = G.subgroups()
    
    normal_subs = []
    
    for H in all_subs:
        is_normal = True
        for g in G:
            l_coset = []
            r_coset = []
            for h in H:
                l_coset.append(g*h)
                r_coset.append(h*g)
                
            l_coset.sort(key=str)
            r_coset.sort(key=str)
            
            if l_coset != r_coset:
                is_normal = False
                break
                
        if is_normal:
            normal_subs.append(H)
            
    actual_count = len(normal_subs)

    # знаходимо дільники n вручну для гіпотези
    divs = divisors(n)
    d_n = len(divs) # кількість дільників
    
    # 2, 3, 4. Перелік порядків та структур
    orders = [H.order() for H in normal_subs]
    structures = [H.structure_description() for H in normal_subs]
    
    combined = sorted(zip(orders, structures), key=lambda x: x[0]) # zip()-зшиває два списка в один
    sorted_orders = [x[0] for x in combined]
    sorted_structures = [x[1] for x in combined]
    
    print(f"n = {n:<2} (Порядок D_{n} = {2*n:<2}) | Знайдено нормальних підгруп: {actual_count}")
    print(f"  Дільники числа {n}: {divs} (кількість дільників = {d_n})")
    print(f"  Порядки підгруп: {sorted_orders}")
    print(f"  Структури: {sorted_structures}")
    
    print("\n")
    print("---------------------------------------")
    print("\n")
```

   $D_n$_?_ 

Кількість нормальних підгруп залежить від кількості дільників числа $n$, яку позначимо як $d(n)$.

- Якщо $n$ - непарне число: Усі осі симетрії фігури однакові (проходять через кут і протилежну сторону). Тому нормальними є лише підгрупи з обертань (їх $d(n)$ штук) та одна повна група. Разом: $d(n)+1$.
- Якщо $n$ - парне число: Осі симетрії бувають двох різних видів (через кути та через середини сторін). Ця відмінність утворює ще $2$ додаткові нормальні підгрупи. Разом: $d(n)+3$.

### 3. Гомоморфізм групи матриць у симетричну групу. 

Розглянемо групу $GL_2(\mathbb{Z}_3)$ та множину прямих $M=\{\ell_1, \ell_2, \ell_3, \ell_4\}$ у просторі $\mathbb{Z}_3^2$, що визначаються парами пропорційних ненульових векторів:
\begin{align*}
\ell_1 = \{ (1,0), (2,0) \}, \
\ell_2 = \{ (0,1), (0,2) \}, \
\ell_3 = \{ (1,1), (2,2) \}, \
\ell_4 = \{ (1,2), (2,1) \}.
\end{align*}

1. Побудуйте групу $GL_2(\mathbb{Z}_3)$.
2. Задайте відображення
   $$\phi: GL_2(\mathbb{Z}_3) \longrightarrow S_4,$$
   яке переводить матрицю у перестановку, що описує її дію на прямі з $M$.
3. Перевірте, що $\phi$ є гомоморфізмом груп.<br>
4. Обчисліть ядро $N=\ker(\phi)$ та образ $\mathrm{Im}(\phi)$.
5. Побудуйте факторгрупу $GL_2(\mathbb{Z}_3)/N$ та порівняйте її з образом $\mathrm{Im}(\phi)$.



```sage
# 1. Будуємо поле Z3 та простір векторів V = Z3^2
F = GF(3)
V = VectorSpace(F, 2)

all_vectors = [v for v in V if v != V.zero_vector()]

l1 = [V([1, 0]), V([2, 0])]
l2 = [V([0, 1]), V([0, 2])]
l3 = [V([1, 1]), V([2, 2])]
l4 = [V([1, 2]), V([2, 1])]

lines = [l1, l2, l3, l4]

print("1. Побуддова множини прямих та групи GL2(Z3):")
print(f"    Множина прямих M: {lines}")

# Матриця оборотна, якщо її визначник != 0
GL2_Z3_matrices = []
for a in F:
    for b in F:
        for c in F:
            for d in F:
                mat = Matrix(F, [[a, b], [c, d]])
                if mat.det() != 0:
                    GL2_Z3_matrices.append(mat)
                    
print(f"    Знайдено елементів у групі GL2(Z3): {len(GL2_Z3_matrices)} (теоретично має бути 48)")

# 2. Відображення phi: GL2(Z3) -> S4
S4 = SymmetricGroup(4)

# Допоміжна функція: застосування матриці до прямої
def apply_matrix_to_line(mat, line):
    new_line = [mat * v for v in line]
    for i, l in enumerate(lines):
        if new_line[0] in l and new_line[1] in l:
            return i + 1
    return None

# Головна функція гомоморфізму
def phi(mat):
    # дивимось, куди матриця перевела кожну з прямих
    img_l1 = apply_matrix_to_line(mat, l1)
    img_l2 = apply_matrix_to_line(mat, l2)
    img_l3 = apply_matrix_to_line(mat, l3)
    img_l4 = apply_matrix_to_line(mat, l4)

    perm_list = [img_l1, img_l2, img_l3, img_l4]
    
    return S4(perm_list)

print("\n2. Тестування відображення phi:")
sample_mat1 = Matrix(F, [[1, 0], [0, 1]])
sample_mat2 = Matrix(F, [[0, 1], [1, 0]])
sample_mat3 = Matrix(F, [[1, 1], [0, 1]])

print(f"    Матриця:\n{sample_mat1}\n  phi(A) = {phi(sample_mat1)}")
print(f"    Матриця:\n{sample_mat2}\n  phi(A) = {phi(sample_mat2)}")
print(f"    Матриця:\n{sample_mat3}\n  phi(A) = {phi(sample_mat3)}")

# 3. Перевірка, що phi - гомоморфізм
print("3. Перевірка на гомоморфізм:")
is_homo = True
for A in GL2_Z3_matrices:
    for B in GL2_Z3_matrices:
        # Оскільки в Sage перестановки перемножуються зліва направо, то змінюємо матриці місцями
        if phi(A * B) != phi(B) * phi(A):
            is_homo = False
            break
    if not is_homo:
        break
        
print(f"    Phi гомоморфізм: {is_homo}")

# 4. Обчислення ядра та образу
print("4. Обчислення ядра N та образу Im:")
id_perm = S4(()) # тотожна перестановка

N_ker = []
Im_phi = []

for mat in GL2_Z3_matrices:
    p = phi(mat)
    if p not in Im_phi:
        Im_phi.append(p)
        
    if p == id_perm:
        N_ker.append(mat)
        
print(f"    Елементи образу: {Im_phi}")
print(f"    Матриці ядра:")
for n in N_ker:
    print(f"{n}\n")
    
# 5. Побудова факторгрупи та порівняння з образом
print("5. Побудова факторгрупи GL2(Z3)/N та порівняння:")

left_cosets = []
for A in GL2_Z3_matrices:
    coset = [A * n for n in N_ker]
    coset.sort(key=str)
    if coset not in left_cosets:
        left_cosets.append(coset)
        
print(f"    Кількість лівих класів суміжності: {len(left_cosets)}")
print(f"    Розмір образу Im(phi): {len(Im_phi)}")

if len(left_cosets) == len(Im_phi):
    print("    -> Розмір факторгрупи збігається з роміром образу!")
else:
    print("    -> Розміри чомусь не збігаються.")
```
