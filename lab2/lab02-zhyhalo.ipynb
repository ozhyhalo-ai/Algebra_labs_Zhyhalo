### 2. Працюємо з елементами груп

Для кожної з груп: $\mathbb{Z}_{10}$, $\mathbb{Z}_{10}^{*}$, $A_4$, $GL_2(\mathbb{Z}_3)$, виконайте наступні кроки:

1. Побудуйте групу.
2. Виберіть довільні елементи $a, b \in G$ та обчисліть добуток $a \cdot b$.
3. Знайдіть обернені елементи $a^{-1}, b^{-1}$ та виконайте перевірку правильності.
4. Побудуйте підгрупу, породжену елементами $a$ та $b$, і визначте її порядок.
5. Знайдіть всі елементи групи, які є оберненими самі до себе, тобто $a^{-1}=a$.
6. Якщо група $G$ не є абелевою, знайдіть такі елементи $a, b \in G$, що $a \cdot b \neq b \cdot a$.



```sage
#Z10
Z = AdditiveAbelianGroup([10])
print("1. Групу побудовано:", Z)
elements = Z.list()

#2.
a = elements[1]
b = elements[2]
print(f"2. Вибрані елементи: a = {a}, b = {b}")
print(f"Сума a+b = {a + b}")

#3.
a_inv = -a
b_inv = -b
print(f"3. Протележний до a: {a_inv}")
print(f"Протележний до b: {b_inv}")

# Перевірка
id_elem = Z.identity()
print(f"Перевірка a + a^-1 == e: {a + a_inv == id_elem}")

#4.
H_elements = set() # використовуємо множину, щоб уникнути дублів

# Перевіряємо всі можливі комбінації додавання a та b
for i in range(Z.order()):
    for j in range(Z.order()):
        H_elements.add(i*a + j*b)
        
print(f"4. Порядок підгрупи, породженої a та b: {len(H_elements)}")
print(f"Елементи підгрупи: {sorted(list(H_elements))}")

#5.
self_inv_elements = [x for x in Z if x == -x]
print(f"5. Елементи, протилежні самі до себе: {self_inv_elements}")

#6.
print("6. Група Z10 є абелевою, тому некомутуючих елементів немає.")
```

```sage
#Z10^*
Z = Integers(10).unit_group()
print("1. Групу побудовано:", Z)
elements = Z.list()

#2.
a = elements[1]
b = elements[2]
print(f"2. Вибрані елементи: a = {a}, b = {b}")
print(f"Добуток a*b = {a * b}")

#3.
a_inv = a^-1 # також працює a.inverse()
b_inv = b^-1
print(f"3. Обернений до a: {a_inv}")
print(f"Обернений до b: {b_inv}")

# Перевірка
id_elem = Z.identity()
print(f"Перевірка a * a^-1 == e: {a * a_inv == id_elem}")

#4.
H = Z.subgroup([a, b])
print(f"4. Порядок підгрупи, породженної a та b: {H.order()}")

#5.
self_inv_elem = [x for x in Z if x.inverse() == x]
print(f"5. Елементи, обрані самі до себе: {self_inv_elem}")

#6.
print("6. Шукаємо елементи, що не комутують:")
if Z.is_abelian():
    print("Група абелева, таких елементів немає.")
else:
    found = False
    for x in Z:
        for y in Z:
            if x * y != y * x:
                print(f"Знайдено: x = {x}, y = {y}")
                print(f"x*y = {x*y}, y*x = {y*x}")
                found = True
                break
        if found:
            break
```

### 3. Підгрупи та їхні властивості

Для кожної з груп $\mathbb{Z}_{24}$, $Q_8$, $D_5$, $S_4$ виконайте наступні кроки:
1. Побудуйте групу.
2. Отримайте та виведіть список усіх підгруп групи.   
3. Визначте кількість підгруп групи з точністю до ізоморфізму.
4. Визначте порядки всіх підгруп групи.
5. Перевірте, чи порядок кожної підгрупи ділить порядок групи.

```sage
#Z24
# (використовуємо ізоморфну циклічну групу перестановок, щоб працювали всі методи)
Z = CyclicPermutationGroup(24)
print("1. Групу побудовано:", Z)

#2.
subgroups = Z.subgroups()
print(f"2. Знайдено підгруп: {len(subgroups)}")
for i, H in enumerate(subgroups):
    print(f"Підгрупа {i+1}: порядок {H.order()}, твірні {H.gens()}")
    
print(f"\n")
    
#3.
iso_classes = set(H.structure_description() for H in subgroups)
print(f"3. З точністю до ізоморфізму підгруп: {len(iso_classes)}")
print(f"Їхні типи (структури): {list(iso_classes)}")

#4.
orders = [H.order() for H in subgroups]
print(f"\n4. Порядки всіх підгруп (відповідно до списку): {orders}")

#5.
group_order = Z.order()
# all() повертає True, якщо умова виконається для всіх порядків
check_div = all(group_order % order == 0 for order in orders)
print(f"\n5. Чи всі порядки ділять {group_order}? {check_div}")
```

### 4. Графи Келі

Графом Келі групи $G$ відносно системи твірних $S$ називається орієнтований граф $\Gamma(G,S)$, вершинами якого є елементи групи і кожний $g\in G$ з'єднаний орієнтованим ребро з $gs$ для всіх $s\in S$.

Для кожної з груп: $D_4$, $Q_8$, $S_3$, $S_4$, $A_4$, виконайте наступні кроки:
1. Побудуйте граф Келі для деякої системи твірних групи.
2. Визначте степінь вершин графа Келі.
3. Перевірте, чи граф Келі є вершинно-транзитивним.
4. Дослідіть, чи граф Келі містить гамільтонів цикл.

```sage
# D4
G = DihedralGroup(4)
print(f"Група: {G}")
print(f"Система твірних S: {G.gens()}")

# Будуємо граф Келі. за замовчуванням Sage будує орієнтований граф
CG = G.cayley_graph()
CG.plot(vertex_labels=True, vertex_size=1000, vertex_color='violet', color_by_label=True).show()
# color_by_label розмальовує ребра відповідному твірному елементу

print(f"\n1. Граф Келі побудовано. Вершин: {CG.order()}, Ребер: {CG.size()}")

#2.
# Оскільки граф Келі регулярний (всі вершини однакови), візьмемо першу
v = list(CG)[0]
out_deg = CG.out_degree(v)
in_deg = CG.in_degree(v)
total_deg = CG.degree(v)
print(f"\n2. Степінь кожної вершини: {total_deg} (вхідний: {in_deg}, вихідний: {out_deg})")

#3.
print(f"\n3.Чи є граф вершино-транзитивним? {CG.is_vertex_transitive()}")

#4.
print(f"\n4. Чи містить граф гамільтонів цикл? {CG.is_hamiltonian()}")
```

### 5*. Ймовірність породження симетричної та знакозмінної груп

1. Створіть групу $S_n$ для деякого $n\geq 10$.
2. Проведіть наступний експеримент $m=1000$ разів:<br>
   2.1. Згенеруйте дві випадкові підстановки $a$ та $b$ в $S_n$.<br>
   2.2. Побудуйте групу $H=\langle a,b\rangle$.<br>
   2.3. Перевірте, чи $H$ дорівнює $S_n$, $H=A_n$ або є іншою підгрупою.
3. Обчисліть частоту кожного результату. 
4. Сформулюйте гіпотезу: яка ймовірність породження $S_n$ та $A_n$ двома підстановками.



```sage
n = 10
G = SymmetricGroup(n)
Sn_order = G.order()
An_order = Sn_order / 2

count_Sn = 0
count_An = 0
count_other = 0
m = 1000

print(f"Починаємо експеримент (m = {m})...")

#2.
for _ in range(m):
    #2.1. генеруємо випадкові підстановки
    a = G.random_element()
    b = G.random_element()
    
    #2.2.
    H = G.subgroup([a, b])
    H_order = H.order()
    
    #2.3.
    if H_order == Sn_order:
        count_Sn += 1
    elif H_order == An_order:
        count_An += 1
    else:
        count_other += 1
        
#3. обчислюємо частоту (ймовірність)
prob_Sn = count_Sn / m
prob_An = count_An / m
prob_other = count_other / m

print("\nРезультати!")
print(f"Згенеровано Sn: {count_Sn} разів (Частота: {prob_Sn*100}%)")
print(f"Згенеровано An: {count_An} разів (Частота: {prob_An*100}%)")
print(f"Згенеровано Іншу: {count_other} разів (Частота: {prob_other*100}%)")
```

Дві випадкові підстановки з високою ймовірністю (близько 89% для n=10) породжують або симетричну групу $S_n$, або альтернуючу $A_n$. При цьому $S_n$ генерується приблизно у 3 рази частіше, ніж $A_n$(співвідношення ~3:1). Зі збільшенням параметра $n$, ймовірність згенерування будь-яку іншу підгрупу буде стрімко прямувати до нуля.
