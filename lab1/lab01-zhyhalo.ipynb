### 1. Арифметичні операції

а) додайте 25 і 17;<br>
б) помножте 7 на 9;<br>
в) поділіть 144 на 12;<br>
г) знайдіть залишок від ділення 156 на 7;<br>
д) поділіть 156 націло на 7;<br>
е) розкладіть число 2024 на прості множники.

```sage
print("a) 25 + 17 =", 25+17)
print("б) 7 * 9 =", 17*9)
print("в) 144 / 12 =", 144/12)
print("г) 156 % 7 =", 156%7)
print("д) 156 // 7 =", 156//7)
print("е) factor(2024)", factor(2024))
```

### 2. Степені, корені та стандартні функції

а) обчисліть $5^3$;<br>
б) знайдіть квадратний корінь з 49;<br>
в) обчисліть $\cos 0$;<br>
г) обчисліть $\sin \pi$;<br>
д) обчисліть $\exp(2)$;<br>
е) обчисліть $\tan\left( \frac{\pi}{4} \right)$.



```sage
print("а) 5^3 =", 5^3) # Можна як **, так і ^
print("б) 49^1/2 =", 49^(1/2)) # Можна ще sqrt(49)
print("в) cos(0) =", cos(0))
print("г) sin(pi) =", sin(pi))
print("д) exp(2) =", exp(2))
print("е) tan(pi/4) =", tan(pi/4))
```

###  3. Символічні обчислення та алгебра виразів
а) оголосіть змінні $x=10$ і $y=15$ та обчисліть їхню суму і добуток;<br>
б) оголосіть символічну змінну $z$ і обчисліть вираз $3z^2 + 2z - 5$;<br>
в) спростіть вираз $2x^2 + 4x - 2x^2 + x$;<br>
г) розкрийте дужки у виразі $(x + 2)(x - 3)$;<br>
д) розкладіть на множники тричлен $x^2 - 9x + 20$;<br>
е) розв'яжіть рівняння $x^2 - 4x + 3 = 0$.<br>

```sage
#a
x, y = 10, 15
print(f"a) {x} + {y} =", x+y)

#б
var('z')
res = 3*z^2+2*z-5
print('б)', res)

#в
var('x') # оголошую x, щоб він знову став символьним
res_2 = 2*x^2 + 4*x - 2*x^2 + x
print('в)', res_2)

#г
print('г)', ((x + 2) * (x - 3)).expand())

#д
expr_d = x^2 - 9*x + 20
print("д)", expr_d.factor())

#е
print("е)", solve(x^2 - 4*x + 3 == 0, x))
```

### 4. Похідні, інтеграли та графіки функцій

а) побудуйте графік функції $f(x) = x^2 - 4x + 3$ на відрізку $[-5, 5]$; <br>
б) знайдіть похідну функції $f(x) = x^3 + 5x^2 - 2x + 7$; <br>
в) знайдіть невизначений інтеграл функції $2x^3 - x^2 + 5$; <br>
г) обчисліть визначений інтеграл $\int_{1}^{4} x^2\, dx$; <br>
д) оголосіть функцію $f(x) = 3x^2 - 2x + 1$ та обчисліть $f(5)$; <br>
е) побудуйте на одній площині графіки $f(x)=x^2$ і $g(x)=-x^2+5$ на $[-3,3]$.


```sage
#a
plot(x^2 - 4*x + 3, (x, -5, 5), color='magenta').show()
# дух октави, але без попереднього генерування масиву точок
# + потестувала додавання кольору
# додала show(), щоб графік не зникав, коли виконуються наступні завдання

#б
res_1 = x^3 + 5*x^2 - 2*x + 7
print("б)", diff(res_1, x))

#в
res_2 = 2*x^3 - x^2 + 5
print("в)", integral(res_2, x))

#г
res_3 = x^2
print("г)", integral(res_3, x, 1, 4))
# це все дуже нагадує октаву

#д
f(x) = 3*x^2 - 2*x + 1
print('д)', f(5))

#е
plot([x^2, -x^2 + 5], (x, -3,3)).show()
# також можна просто намалювати два графіка окремо і додати
```

а) напишіть умовний оператор, який визначає, чи число $n$ додатне, від'ємне або дорівнює нулю; <br>
